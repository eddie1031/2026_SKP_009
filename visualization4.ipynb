```python
heatmap_df = pd.DataFrame(
    [
        [120, 150, 180, 210],
        [90, 110, 130, 160],
        [70, 95, 105, 125],
    ],
    index=[
        "PC",
        "모바일",
        "주변기기",
    ],
    columns=[
        "1월",
        "2월",
        "3월",
        "4월",
    ],
)

print(heatmap_df)
```

```python
fig = px.imshow(
    heatmap_df,
    text_auto=True,
    aspect='auto',
    labels={
        'x': '월',
        'y': '카테고리',
        'color': '매출'
    }
)

fig.show()
```

```python
fig = px.imshow(
    heatmap_df,
    text_auto=True,
    aspect='auto',
    labels={
        'x': '월',
        'y': '카테고리',
        'color': '매출'
    },
    color_continuous_scale='Blues'
)

fig.show()
```

```python
numeric_df = product_df[
    [
        'sales',
        'profit',
        'customers',
        'satisfaction'
    ]
]

numeric_df.head()
```

```python
corr_df = numeric_df.corr().round(2)

corr_df
```

상관계수 행렬

- 양수 1에 가까울수록: 강한 양의 선형 관계
- 0에 가까울수록: 선형 관계가 약하다
- 음수 1에 가까울수록: 강한 음의 선형 관계


```python
fig = px.imshow(
    corr_df,
    text_auto='.2f',
    zmin=-1,
    zmax=1,
    color_continuous_scale='RdBu_r'
)

fig.show()
```

```python
fig = px.treemap(
    product_df,
    path=[
        'department',
        'category',
        'product'
    ],
    values='sales',
    title='상품 계층별 매출'
)

fig.show()
```

```python
fig = px.treemap(
    product_df,
    path=[
        'department',
        'category',
        'product'
    ],
    values='sales',
    color='profit',
    title='상품 계층별 매출'
)

fig.show()
```

```python
fig = px.sunburst(
    product_df,
    path=[
        'department',
        'category',
        'product'
    ],
    values='sales',
)

fig.show()
```

```python
fig = px.sunburst(
    product_df,
    path=[
        'department',
        'category',
        'product'
    ],
    color='profit',
    values='sales',
)

fig.show()
```

```python
fig = px.scatter_3d(
    product_df,
    x='sales',
    y='profit',
    z='customers',
    hover_name='product',
    title='매출, 이익, 고객 수의 관계'
)

fig.show()
```

```python
fig = px.scatter_3d(
    product_df,
    x='sales',
    y='profit',
    z='customers',
    color='category',
    hover_name='product',
    title='매출, 이익, 고객 수의 관계'
)

fig.show()
```
