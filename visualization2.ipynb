```python
import pandas as pd
import plotly.express as px

df = pd.DataFrame({
    "product": [
        "노트북", "모니터", "키보드", "마우스",
        "태블릿", "헤드셋", "웹캠", "스피커",
        "SSD", "공유기", "스마트폰", "충전기",
    ],
    "category": [
        "PC", "PC", "주변기기", "주변기기",
        "모바일", "주변기기", "주변기기", "주변기기",
        "PC", "네트워크", "모바일", "모바일",
    ],
    "ad_cost": [
        28, 18, 10, 7,
        22, 9, 6, 8,
        13, 5, 30, 4,
    ],
    "sales": [
        320, 210, 150, 110,
        260, 140, 90, 120,
        180, 100, 360, 80,
    ],
    "customers": [
        120, 95, 150, 170,
        105, 130, 75, 100,
        115, 60, 140, 180,
    ],
    "profit": [
        72, 48, 41, 28,
        58, 35, 21, 26,
        46, 22, 81, 18,
    ],
})

print(df)
```
