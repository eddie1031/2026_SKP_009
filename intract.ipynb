```python
fig = px.scatter(
    product_df,
    x='sales',
    y='profit',
)

fig.show()
```

```python
fig = px.scatter(
    product_df,
    x='sales',
    y='profit',
    hover_name='product'
)

fig.show()
```

```python
fig = px.scatter(
    product_df,
    x='sales',
    y='profit',
    hover_name='product',
    hover_data=[
        'category',
        'region',
        'customers',
        'satisfaction',
    ]
)

fig.show()
```

```python
fig = px.scatter(
    product_df,
    x='sales',
    y='profit',
    hover_name='product',
    hover_data={
        'category': True,
        'region': True,
        'customers': ":,.0f",
        'satisfaction': ":.1f",
    }
)

fig.show()
```

```python
fig = px.scatter(
    product_df,
    x='sales',
    y='profit',
    color='category',
    hover_name='product',
)

fig.show()
```

```python
fig = px.scatter(
    product_df,
    x='sales',
    y='profit',
    color='category',
    hover_name='product',
)

fig.update_layout(
    legend={
        'orientation': 'h',
    }
)

fig.show()
```

```python
fig = px.scatter(
    product_df,
    x='sales',
    y='profit',
    color='category',
    size='customers',
    hover_name='product',
    hover_data={
        'region': True,
        'satisfaction': ':.1f',
    }
)

fig.show(
    config={
        'scrollZoom': True,
    }
)
```
