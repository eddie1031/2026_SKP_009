```python
import pandas as pd
import plotly.express as px

df = pd.DataFrame({
    "group": (
        ["A"] * 12
        + ["B"] * 12
        + ["C"] * 12
    ),
    "score": [
        # A 그룹
        52, 55, 58, 60, 61, 63,
        65, 67, 68, 70, 72, 95,

        # B 그룹
        48, 50, 53, 54, 56, 57,
        58, 60, 62, 64, 66, 68,

        # C 그룹
        60, 62, 64, 66, 68, 70,
        72, 74, 76, 78, 80, 82,
    ],
})

print(df)
```

```python
df.groupby('group')['score'].describe()
```

```python
fig = px.histogram(
    df,
    x='score',
    y='group',
)

fig.show()
```

```python
px.histogram(
    df,
    x='score',
    nbins=5,
)
```

```python
df['group'].value_counts()
```

```python
fig = px.histogram(
    df,
    x='score',
    color='group',
    opacity=0.6,
    # barmode='overlay'
)

fig.show()
```

```python
fig = px.histogram(
    df, # 데이터
    x='score', # X축
    color='group',  # 그룹핑(색상으로 구분)
    opacity=0.6, # 투명도
    barmode='overlay' # 막대기 유형
)

fig.show()
```

```python
# A그룹에 대한 상자수염그래프를 그리시오
group_a = df[df['group'] == 'A']

group_a
```

```python
fig = px.box(
    group_a,
    y='score',
    title='A그룹 점수별 분포'
)

fig.show()
```

```python
fig = px.box(
    df,
    x='group',
    y='score',
)

fig.show()
```

```python
fig = px.box(
    df,
    x='group',
    y='score',
    color='group',
)

fig.show()
```

```python
fig = px.violin(
    group_a,
    y='score',
)

fig.show()
```

```python
fig = px.violin(
    group_a,
    y='score',
    points='all',
)

fig.show()
```

```python
fig = px.violin(
    df,
    x='group',
    y='score',
    color='group',
    points='all',
    title='그룹별 점수 밀도'
)

fig.show()
```
