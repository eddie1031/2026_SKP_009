````python
import pandas as pd
import plotly.express as px

monthly_df = pd.DataFrame({
    "date": pd.date_range(
        start="2026-01-01",
        periods=12,
        freq="MS",
    ),
    "sales": [
        120, 135, 128, 150,
        165, 172, 168, 185,
        198, 210, 205, 225,
    ],
})

print(monthly_df)
````

````python
fig = px.line(
    monthly_df,
    x='date',
    y='sales',
)

fig.show()
````

````python
monthly_df['change'] = monthly_df['sales'].diff()
````

````python
monthly_df
````

````python
unsorted_df = pd.DataFrame({
    "date": pd.to_datetime([
        "2026-01-01",
        "2026-03-01",
        "2026-02-01",
        "2026-05-01",
        "2026-04-01",
    ]),
    "sales": [
        120,
        128,
        135,
        165,
        150,
    ],
})

print(unsorted_df)
````

````python
fig = px.line(
    unsorted_df,
    x='date',
    y='sales',
    markers=True,
)

fig.show()
````

````python
sorted_df = unsorted_df.sort_values('date')

sorted_df
````

````python
fig = px.line(
    sorted_df,
    x='date',
    y='sales',
    markers=True,
)

fig.show()
````

````python
category_df = pd.DataFrame({
    "date": pd.to_datetime([
        "2026-01-01", "2026-02-01", "2026-03-01",
        "2026-04-01", "2026-05-01", "2026-06-01",

        "2026-01-01", "2026-02-01", "2026-03-01",
        "2026-04-01", "2026-05-01", "2026-06-01",

        "2026-01-01", "2026-02-01", "2026-03-01",
        "2026-04-01", "2026-05-01", "2026-06-01",
    ]),
    "category": (
        ["PC"] * 6
        + ["모바일"] * 6
        + ["주변기기"] * 6
    ),
    "sales": [
        180, 195, 205, 220, 235, 250,
        140, 150, 165, 180, 205, 225,
        100, 108, 115, 125, 130, 142,
    ],
})

print(category_df)
````

````python
fig = px.line(
    category_df,
    x='date',
    y='sales',
    markers=True,
)

fig.show()
````

````python
fig = px.line(
    category_df,
    x='date',
    y='sales',
    color='category',
    markers=True,
)

fig.show()
````

````python
fig = px.line(
    category_df,
    x='date',
    y='sales',
    color='category',
    symbol='category',
    markers=True,
)

fig.show()
````
