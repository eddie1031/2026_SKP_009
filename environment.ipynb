```python
import pandas as pd
import plotly.express as px

df = pd.DataFrame({
    'product': ['노트북','모니터','키보드','마우스'],
    'sales': [12,17,10,9]
})

df
```

```python
fig = px.bar(
    df,
    x='product',
    y='sales',
    title='제품별 매출'
)

fig.show()
```

```python
df = pd.DataFrame({
    'month': ['1월','2월','3월','4월','5월', '6월',],
    'sales': [120,130,140,128,160,150]
})

fig = px.line(
    df,
    x='month',
    y='sales',
    markers=True,
    title='월별 매출 변화량'
)

fig.show()
# mode bar
```

```python
df = pd.DataFrame({
    'month': [
        '1월', '1월',
        '2월', '2월',
        '3월', '3월',
    ],
    'category': [
        'PC', '주변기기',
        'PC', '주변기기',
        'PC', '주변기기',
    ],
    'sales': [
        120, 80,
        140, 95,
        160, 110,
    ]
})

fig = px.bar(
    df,
    x='month',
    y='sales',
    color='category',
    barmode='group',
    title='월별 카테고리 매출'
)

fig.show()
```

## 데이터 시각화 툴 비교

|구분|Matplotlib|Seaborn|Plotly|
|---|---|---|---|
|목적|범용 그래프 작성|통계 데이터 시각화|인터렉티브 시각화|
|웹대시보드활용|추가작업이 필요|추가작업이 필요|바로 활용 가능|
|세부제어|가장 세밀하게 제어가능|Matplotlib을 사용|Graph Object 활용|
